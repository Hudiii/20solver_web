```python
import pandas as pd

file_path = '/content/drive/MyDrive/accepted_2007_to_2018Q4.csv'
cols_needed = ['funded_amnt', 'total_pymnt', 'loan_status']

df = pd.read_csv(file_path, usecols=cols_needed)

print("=== Jumlah baris total ===")
print(len(df))

print("\n=== Missing value per kolom ===")
print(df.isnull().sum())

print("\n=== Distribusi loan_status (semua kategori) ===")
print(df['loan_status'].value_counts())
```
