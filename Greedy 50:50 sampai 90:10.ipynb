```python
df = pd.read_csv("/content/drive/MyDrive/Dataset_Wisata_Solo.csv")

#Konversi tipe data agar cost = int, rating = float
df["Biaya"] = (df["Biaya"]/1000).astype(int)
df["Rating"] = df["Rating"].astype(float)

#Ubah ke format list of tuples: (Nama, Biaya, Rating)
places = [(row["Nama"], int(row["Biaya"]), float(row["Rating"]))
          for _, row in df.iterrows()]
```

# Preprocessing

```python
free_places = [p for p in places if p[1] == 0]
paid_places = [p for p in places if p[1] > 0]
```

##Algoritma Greedy

```python
def greedy_with_weights(places, max_cost, w_rating, w_ratio):
    def score(p):
        name, cost, rating = p
        if cost == 0:
            return float('inf')
        return w_rating * rating + w_ratio * (rating / cost)

    data = sorted(places, key=score, reverse=True)

    total_rating = 0
    total_cost = 0
    selected = []

    for name, cost, rating in data:
        if total_cost + cost <= max_cost:
            total_cost += cost
            total_rating += rating
            selected.append(name)

    return total_rating, selected


def greedy_wide_table(places):
    weights = [(0.5, 0.5), (0.6, 0.4), (0.7, 0.3), (0.8, 0.2), (0.9, 0.1)]

    print("\n===== GREEDY RATING WIDE TABLE =====")
    print(f"{'Budget':<10} {'50:50':<10} {'60:40':<10} {'70:30':<10} {'80:20':<10} {'90:10':<10}")
    print("-" * 70)

    for budget in range(10000, 1000001, 10000):
        budget_ribu = budget // 1000
        ratings = []

        for w_rating, w_ratio in weights:
            rating, _ = greedy_with_weights(places, budget_ribu, w_rating, w_ratio)
            ratings.append(f"{rating:.2f}")

        print(f"{budget:<10} {ratings[0]:<10} {ratings[1]:<10} {ratings[2]:<10} {ratings[3]:<10} {ratings[4]:<10}")


# Jalankan
greedy_wide_table(paid_places)
```

```python
def greedy_best_for_each_budget(places):
    weights = [(0.5, 0.5), (0.6, 0.4), (0.7, 0.3), (0.8, 0.2), (0.9, 0.1)]

    print("\n===== GREEDY BEST RESULT PER BUDGET =====")
    print(f"{'Budget':<10} {'Best Rating':<12} {'Best Bobot':<12} {'Destinations'}")
    print("-" * 120)

    for budget in range(10000, 1000001, 10000):
        budget_ribu = budget // 1000

        best_rating = -1
        best_weight = None
        best_dest = None

        for w_rating, w_ratio in weights:
            rating, selected = greedy_with_weights(places, budget_ribu, w_rating, w_ratio)

            if rating > best_rating:
                best_rating = rating
                best_weight = f"{int(w_rating*100)}:{int(w_ratio*100)}"
                best_dest = selected

        print(f"{budget:<10} {best_rating:<12.2f} {best_weight:<12} {best_dest}")

greedy_best_for_each_budget(paid_places)
```
